## 1. KASUS 1: METODE GEOLISTRIK RESISTIVITAS 2D (AKUIFER & BEDROCK)
*Berkas Rujukan:* [01_interpretasi_geolistrik.md](file:///C:/Users/Lenovo/.gemini/antigravity/scratch/geofisika_unhas_ai/01_interpretasi_geolistrik.md)

### A. Input Data & Parameter Survei
* **Lokasi & Geologi:** Pesisir Barombong - Somba Opu (Makassar/Gowa). Endapan Aluvium Kuarter (Qac) menindih Formasi Batuan Gunungapi Camba (Tmc).
* **Konfigurasi & Bentang:** Wenner-Schlumberger, panjang lintasan 200 m (spasi elektroda 5 m), arah Barat Daya - Timur Laut (SW - NE).
* **Data Inversi (Res2Dinv):** Iterasi ke-5, *absolute error* $3.2\%$, rentang resistivitas $2.5 - 580\ \Omega\cdot\text{m}$.

### B. Hasil Interpretasi Sesuai Struktur Output [O]
1. **Zonasi Resistivitas & Litologi:**
   * **Zona I ($18 - 40\ \Omega\cdot\text{m}$, kedalaman $0 - 2.5\text{ m}$):** Lapisan penutup lapukan (*topsoil*) dan lempung pasiran tak jenuh (*vadose zone*).
   * **Zona II ($2.5 - 8\ \Omega\cdot\text{m}$, kedalaman $2.5 - 12\text{ m}$, bentang $0 - 70\text{ m}$):** Lempung marin jenuh air payau/salin akibat rembesan air laut pesisir (*coastal saline intrusion*).
   * **Zona III ($45 - 95\ \Omega\cdot\text{m}$, kedalaman $12 - 38\text{ m}$, bentang $60 - 175\text{ m}$):** Pasir kasar berseling kerikil aluvial jenuh air tawar (*confined freshwater aquifer*).
   * **Zona IV ($220 - 580\ \Omega\cdot\text{m}$, kedalaman $> 28 - 38\text{ m}$):** Batuan dasar (*bedrock*) vulkanik masif Formasi Camba (breksi/batupasir kompak).
2. **Karakteristik Akuifer:**
   * **Kedalaman:** Top pada $12 - 14\text{ m}$, Bottom pada $35 - 38\text{ m}$, tebal bersih $\approx 22\text{ meter}$.
   * **Kondisi:** Akuifer tertekan yang dilindungi lapisan lempung tebal di atasnya.
3. **Posisi Batuan Dasar (*Bedrock*):** Berada pada kedalaman $> 38\text{ m}$ di bagian barat daya, dan terangkat (*bedrock high*) hingga kedalaman $28\text{ m}$ di timur laut ($x = 140 - 200\text{ m}$).
4. **Rekomendasi Pengeboran Air Tanah:**
   * **Titik Bor:** Jarak bentang horizontal $x = 110\text{ m}$ hingga $x = 130\text{ m}$.
   * **Kedalaman Target:** $35\text{ meter}$.
   * **Konstruksi:** Saringan *screen* pada $18 - 32\text{ m}$ disertai *grouting sanitary seal* pada $0 - 14\text{ m}$ guna mencegah perkolasi air asin dangkal.

## 2. KASUS 2: METODE SEISMIK REFRAKSI DANGKAL (INTERCEPT TIME METHOD)
*Berkas Rujukan:* [02_interpretasi_seismik_refraksi.md](file:///C:/Users/Lenovo/.gemini/antigravity/scratch/geofisika_unhas_ai/02_interpretasi_seismik_refraksi.md)

### A. Input Data & Pengamatan Waktu Tiba
* 12 Geofon (spasi 5 m, $x = 5$ s.d. $60\text{ m}$), *shotpoint* forward pada $x = 0\text{ m}$.
* Gelombang Langsung (G1–G4, $x = 5 - 20\text{ m}$): $t = 10.0, 20.0, 30.0, 40.0\text{ ms}$.
* Gelombang Terbias (G5–G12, $x = 25 - 60\text{ m}$): $t = 44.5, 47.0, 49.5, 52.0, 54.5, 57.0, 59.5, 62.0\text{ ms}$.

### B. Proses Perhitungan Bertahap
1. **Kemiringan Garis ($m = \Delta t / \Delta x$):**
   $$m_1 = \frac{40.0 - 10.0}{20.0 - 5.0} = 2.0\text{ ms/m} = 0.002000\text{ s/m}$$
   $$m_2 = \frac{62.0 - 44.5}{60.0 - 25.0} = 0.5\text{ ms/m} = 0.000500\text{ s/m}$$
2. **Kecepatan Lapisan:**
   $$V_1 = \frac{1}{m_1} = \frac{1}{0.002} = \mathbf{500.0\text{ m/s}}$$
   $$V_2 = \frac{1}{m_2} = \frac{1}{0.0005} = \mathbf{2000.0\text{ m/s}}$$
3. **Waktu Intercept ($t_i$):**
   $$t_i = t_2(x) - m_2 \cdot x = 47.0 - (0.5 \times 30.0) = \mathbf{32.0\text{ ms}} = \mathbf{0.0320\text{ s}}$$
4. **Kedalaman Interface ($z$):**
   $$z = \frac{t_i}{2} \frac{V_1 \cdot V_2}{\sqrt{V_2^2 - V_1^2}} = \frac{0.032}{2} \frac{500 \times 2000}{\sqrt{2000^2 - 500^2}} = 0.016 \times 516.398 = \mathbf{8.26\text{ meter}}$$
5. **Interpretasi Geoteknik:** Lapisan 1 ($500\text{ m/s}$) merupakan tanah lapukan lepas (*soil overburden*), sedangkan lapisan 2 ($2000\text{ m/s}$) merupakan batuan dasar keras (*engineering bedrock*). Fondasi tiang pancang gedung disarankan bertumpu pada kedalaman $> 9.5\text{ m}$.

## 3. KASUS 3: METODE GEOMAGNETIK (TOTAL MAGNETIC ANOMALY & RTP)
*Berkas Rujukan:* [03_interpretasi_geomagnetik.md](file:///C:/Users/Lenovo/.gemini/antigravity/scratch/geofisika_unhas_ai/03_interpretasi_geomagnetik.md)

### A. Input Data & Parameter Geomagnetik
* **Lokasi:** Zona Sesar Walanae / Camba, Sulawesi Selatan.
* **IGRF:** $F_0 = 41\,250\text{ nT}$, Inklinasi $I = -23.4^\circ$, Deklinasi $D = +1.15^\circ$.
* **Anomali Magnetik Total (TMA):** $-385.4\text{ nT}$ (utara) hingga $+245.8\text{ nT}$ (selatan).
* **Hasil Reduksi ke Kutub (RTP):** Monopolar terpusat $+482.0\text{ nT}$.

### B. Sintesis Interpretasi
1. **Penyebab Pola Dipolar:** Inklinasi medan bumi bernilai negatif di belahan bumi selatan dekat khatulistiwa menyebabkan vektor medan terinduksi condong ke atas, menghasilkan kutub negatif di utara dan positif di selatan.
2. **Efektivitas RTP:** Menghilangkan kemiringan arah medan induksi, mengoreksi pergeseran lateral sebesar $850\text{ m}$ ke utara, dan menempatkan anomali positif maksimum (+482 nT) tepat di atas bodi batuan sumber.
3. **Delineasi Geologi:**
   * **Intrusi Batuan Beku:** Bodi intrusif andesit/basaltik Formasi Camba yang kaya mineral magnetit ($\kappa \approx 0.05\text{ SI}$).
   * **Kelurusan Struktur Sesar:** Ditemukan *magnetic lineament* berarah NW - SE sepanjang $> 4\text{ km}$ yang merefleksikan zona patahan aktif sistem Sesar Walanae.

## 4. KASUS 4: SEISMOLOGI TEKTONIK (ANALISIS FOCAL MECHANISM GEMPA BUMI)
*Berkas Rujukan:* [04_interpretasi_focal_mechanism.md](file:///C:/Users/Lenovo/.gemini/antigravity/scratch/geofisika_unhas_ai/04_interpretasi_focal_mechanism.md)

### A. Solusi Nodal Plane & Sumbu Utama
* **Gempa Kerak Dangkal:** $M_w\ 6.1$, Kedalaman $h = 10\text{ km}$, Kawasan Tektonik Sulawesi.
* **Nodal Plane 1 (NP1):** Strike = $350^\circ$, Dip = $82^\circ$, Rake = $-15^\circ$.
* **Nodal Plane 2 (NP2):** Strike = $84^\circ$, Dip = $75^\circ$, Rake = $-171^\circ$.
* **Sumbu Kompresi (P-Axis):** Azimuth $37^\circ$ (NNE-SSW), Plunge $5^\circ$ (sub-horizontal).
* **Sumbu Ekstensi (T-Axis):** Azimuth $128^\circ$ (SE-NW), Plunge $12^\circ$ (sub-horizontal).

### B. Mekanisme & Konteks Geodinamika
1. **Tipe Mekanisme Sesar:** **Sesar Geser Mendatar Mengiri (*Sinistral Strike-Slip*)** dengan komponen normal (*transtensional*) minor.
2. **Pemecahan Ambiguitas Nodal Plane:** Sebaran gempa susulan (*aftershocks*) dan kelurusan gawir sesar permukaan berarah NNW-SSE membuktikan secara definitif bahwa **NP1 adalah Bidang Sesar Aktual (*Fault Plane*)** dan NP2 adalah Bidang Bantu (*Auxiliary Plane*).
3. **Rezim Tektonik:** Karakteristik khas sistem patahan megageser aktif Sulawesi (misal segmen Sesar Palu-Koro atau Sesar Majene) akibat akumulasi gaya konvergensi lempeng di Indonesia Timur.
